### 08) WAP to create an UDF named frequency to count occurances of the specific word in a given text file.

```python
def frequency(word, filepath):
    with open(filepath) as fp:
        l1 = fp.read().split()
        count = l1.count(word)
        print(f'Count of the {word} is {count}.')

frequency('Engineering', './Lab 9 Files/F3.txt')
```

### 09) WAP to get the score of five subjects from the user, store them in a file. Fetch those marks and find the highest score.

```python
def get_scores():
    scores = []
    for i in range(1, 6):
        score = int(input(f"Enter the score for subject {i}: "))
        scores.append(score)
        
    with open('./Lab 9 Files/F5.txt', 'w') as fp:
        for score in scores:
            fp.write(f"{score}\n")

def find_highest_score():
    with open('./Lab 9 Files/F5.txt') as fp:
        scores = [int(line.strip()) for line in fp.readlines()]
        
    highest_score = max(scores)
    print(f"The highest score is: {highest_score}")

get_scores()
find_highest_score()
```

### 10) WAP to write first 100 prime numbers to a file named primenumbers.txt 
(Note: each number should be in new line)

```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

def write_prime_numbers_to_file():
    primes = []
    num = 2
    while len(primes) < 100:
        if is_prime(num):
            primes.append(num)
        num += 1

    with open('./Lab 9 Files/primenumbers.txt', 'w') as fp:
        for prime in primes:
            fp.write(f"{prime}\n")
    print('Added first 100 prime numbers to "primenumbers.text".')

write_prime_numbers_to_file()
```

### 11) WAP to merge two files and write it in a new file.

```python
def merge_files(file1, file2, merged_file):
    f1 = open(file1, 'r')
    f2 = open(file2, 'r')
    output = open(merged_file, 'w')
    
    output.write(f1.read())
    output.write("\n")
    output.write(f2.read())
    
    f1.close()
    f2.close()
    output.close()

    print(f"Contents of {file1} and {file2} have been merged into {merged_file}")

file1 = './Lab 9 Files/file1.txt'
file2 = './Lab 9 Files/file2.txt'
merged_file = './Lab 9 Files/merged_file.txt'

merge_files(file1, file2, merged_file)
```

### 12) WAP to replace word1 by word2 of a text file. Write the updated data to new file.

```python
def replace_word_in_file(input_file, output_file, word1, word2):
    infile = open(input_file, 'r')
    content = infile.read()
    infile.close()

    updated_content = content.replace(word1, word2)

    outfile = open(output_file, 'w')
    outfile.write(updated_content)
    outfile.close()

    print(f"Replaced '{word1}' with '{word2}' and saved to {output_file}")

input_file = './Lab 9 Files/input.txt'
output_file = './Lab 9 Files/output.txt'
word1 = 'oldword'
word2 = 'newword'

replace_word_in_file(input_file, output_file, word1, word2)
```
