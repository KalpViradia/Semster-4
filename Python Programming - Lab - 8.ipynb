### 01) Write a function to calculate BMI given mass and height. (BMI = mass/h**2)

```python
def calculate_bmi(mass, height):
    bmi = mass / (height ** 2)
    return bmi

mass = 70
height = 1.75

bmi = calculate_bmi(mass, height)
print(f"BMI: {bmi}")
```

### 03) Write a function that returns 1 if the given number is Prime or 0 otherwise.

```python
def is_prime(number):
    if number <= 1:
        return 0
    for i in range(2, int(number ** 0.5) + 1):
        if number % i == 0:
            return 0
    return 1

num = 5
result = is_prime(num)
print(f"Is {num} prime? {result}")
```

### 04) Write a function that returns the list of Prime numbers between given two numbers.

```python
def is_prime(number):
    if number <= 1:
        return False
    for i in range(2, int(number ** 0.5) + 1):
        if number % i == 0:
            return False
    return True

def primes_between(start, end):
    prime_numbers = []
    for num in range(start, end + 1):
        if is_prime(num):
            prime_numbers.append(num)
    return prime_numbers

start = 10
end = 50
prime_numbers = primes_between(start, end)
print(f"Prime numbers between {start} and {end}: {prime_numbers}")
```

### 05) Write a function that returns True if the given string is Palindrome or False otherwise.

```python
def is_palindrome(s):
    cleaned_s = ''.join(e for e in s.lower() if e.isalnum())
    return cleaned_s == cleaned_s[::-1]

s = "level"
result = is_palindrome(s)
print(f"Is the string '{s}' a palindrome? {result}")
```

### 08) Write a recursive function to find nth term of Fibonacci Series.

```python
def fibonacci(n):
    if n <= 1:
        return n
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)

n = 5
result = fibonacci(n)
print(result)
```

### 10) Write a function to get the sum of the scores ending with zero.
#### Example : scores = [200, 456, 300, 100, 234, 678] 
#### Ans = 200 + 300 + 100 = 600

```python
def sum_scores_ending_with_zero(scores):
    return sum(score for score in scores if score % 10 == 0)

scores = [200, 456, 300, 100, 234, 678]
result = sum_scores_ending_with_zero(scores)
print(result)
```

### 11) Write a function to invert a given Dictionary.
#### hint: keys to values & values to keys
#### Before : {'a': 10, 'b':20, 'c':30, 'd':40}
#### After : {10:'a', 20:'b', 30:'c', 40:'d'}

```python
def invert_dictionary(original_dict):
    return {v: k for k, v in original_dict.items()}

original_dict = {'a': 10, 'b': 20, 'c': 30, 'd': 40}
inverted_dict = invert_dictionary(original_dict)
print(inverted_dict)
```

### 12) Write a function to check whether the given string is Pangram or not.
#### hint: Pangram is a string containing all the characters a-z atlest once.
#### "the quick brown fox jumps over the lazy dog" is a Pangram string.

```python
def is_pangram(string):
    string = string.lower()
    alphabet = set('abcdefghijklmnopqrstuvwxyz')
    return set(string) >= alphabet

sentence = "the quick brown fox jumps over the lazy dog"
result = is_pangram(sentence)
print(result)
```

### 13) Write a function that returns the number of uppercase and lowercase letters in the given string. 
#### example : Input : s1 = AbcDEfgh ,Ouptput : no_upper = 3, no_lower = 5

```python
def count_upper_lower(s):
    no_upper = sum(1 for char in s if char.isupper())
    no_lower = sum(1 for char in s if char.islower())
    return no_upper, no_lower

s1 = "AbcDEfgh"
no_upper, no_lower = count_upper_lower(s1)
print(f"no_upper = {no_upper}, no_lower = {no_lower}")
```

### 15) For the given list of names of students, extract the names having more that 7 characters. Use filter().

```python
def filter_long_names(names):
    return list(filter(lambda name: len(name) > 7, names))

students = ["John", "Alexander", "Sophia", "Michael", "Emma", "Jonathan"]
long_names = filter_long_names(students)
print(long_names)
```

### 16) For the given list of names of students, convert the first letter of all the names into uppercase. use map().

```python
def capitalize_names(names):
    return list(map(lambda name: name.capitalize(), names))

students = ["john", "alexander", "sophia", "michael", "emma", "jonathan"]
capitalized_names = capitalize_names(students)
print(capitalized_names)
```

### 17) Write udfs to call the functions with following types of arguments:
1. Positional Arguments
2. Keyword Arguments
3. Default Arguments
4. Variable Legngth Positional(*args) & variable length Keyword Arguments (**kwargs)
5. Keyword-Only & Positional Only Arguments

```python
# 1. Positional Arguments
def positional_args(arg1, arg2, arg3):
    print(f"arg1: {arg1}, arg2: {arg2}, arg3: {arg3}")

# 2. Keyword Arguments
def keyword_args(arg1, arg2, arg3):
    print(f"arg1: {arg1}, arg2: {arg2}, arg3: {arg3}")

# 3. Default Arguments
def default_args(arg1, arg2=20, arg3=30):
    print(f"arg1: {arg1}, arg2: {arg2}, arg3: {arg3}")

# 4. Variable Length Positional (*args) & Variable Length Keyword (**kwargs)
def variable_length_args(*args, **kwargs):
    print(f"Positional arguments: {args}")
    print(f"Keyword arguments: {kwargs}")

# 5. Keyword-Only and Positional-Only Arguments
def mixed_args(arg1, arg2, /, arg3, arg4, *, arg5):
    print(f"Positional-only arguments: {arg1}, {arg2}")
    print(f"Positional-or-keyword argument: {arg3}, {arg4}")
    print(f"Keyword-only argument: {arg5}")

positional_args(10, 20, 30)
keyword_args(arg1=10, arg2=20, arg3=30)
default_args(10)
default_args(10, 40)
variable_length_args(10, 20, 30, name="John", age=25)
mixed_args(10, 20, 30, 40, arg5=50)
```
