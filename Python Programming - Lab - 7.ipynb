# Set & Dictionary

### 01) WAP to iterate over a set.

```python
s1 = {1, 2, 3, 4, 5}

for i in s1:
    print(i)
```

### 02) WAP to convert set into list, string and tuple.

```python
s1 = {1, 2, 3, 4, 5}

l1 = list(s1)
print(f'List: {l1}')

str1 = ''.join(map(str, s1))
print(f'String: {str1}')

t1 = tuple(s1)
print(f'Tuple: {t1}')
```

### 03) WAP to find Maximum and Minimum from a set.

```python
s1 = {1, 2, 3, 4, 5}

print(f'Maximum: {max(s1)}')
print(f'Minimum: {min(s1)}')
```

### 04) WAP to perform union of two sets.

```python
s1 = {1, 2, 3, 4, 5}
s2 = {3, 4, 5, 6, 7}

unionSet = s1 | s2 # or s1.union(s2)
print(unionSet)
```

### 05) WAP to check if two lists have at-least one element common.

```python
l1 = [1, 2, 3, 4]
l2 = [4, 5, 6, 7]

commonElements = set(l1) & set(l2)

if commonElements:
    print(f'The lists have at least one element in common: {commonElements}')
else:
    print('The lists do not have any common elements.')
```

### 09) WAP to check whether two given strings are anagram or not using set.

```python
str1 = 'listen'
str2 = 'silent'

str1 = str1.replace(' ', '').lower()
str2 = str2.replace(' ', '').lower()

if set(str1) == set(str2) and len(str1) == len(str2):
    print(f"'{str1}' and '{str2}' are anagrams.")
else:
    print(f"'{str1}' and '{str2}' are not anagrams.")
```

### 10) WAP to find common elements in three lists using set.

```python
list1 = [1, 2, 3, 4, 5]
list2 = [3, 4, 5, 6, 7]
list3 = [4, 5, 8, 9]

set1 = set(list1)
set2 = set(list2)
set3 = set(list3)

commonElements = set1 & set2 & set3
print(f'Common elements in the three lists are: {commonElements}')
```

### 11) WAP to count number of vowels in given string using set.

```python
str1 = 'Hello, how are you?'

vowels = {'a', 'e', 'i', 'o', 'u'}
vowelCount = 0
for char in str1.lower():
    if char in vowels:
        vowelCount += 1

print(f'Number of vowels in the string are: {vowelCount}')
```

### 12) WAP to check if a given string is binary string or not.

```python
str1 = '1010101'

if set(str1).issubset({'0', '1'}):
    print(f"'{str1}' is a binary string.")
else:
    print(f"'{str1}' is not a binary string.")
```

### 13) WAP to sort dictionary by key or value.

```python
d1 = {'apple': 5, 'banana': 2, 'cherry': 7, 'date': 1}

sortByKey = dict(sorted(d1.items()))
print(f'Dictionary sorted by key: {sortByKey}')

sortByValue = dict(sorted(d1.items(), key = lambda item: item[1]))
print(f'Dictionary sorted by value: {sortByValue}')
```

### 14) WAP to find the sum of all items (values) in a dictionary given by user. (Assume: values are numeric)

```python
d1 = {}

n = int(input('Enter the number of items in the dictionary: '))

for _ in range(n):
    key = input('Enter key: ')
    value = float(input(f'Enter value for {key}: '))
    d1[key] = value

totalSum = sum(d1.values())
print(f'The sum of all values in the dictionary is: {totalSum}')
```
